```python
import pandas as pd
import traci
import traceback
import sumolib

class SumoSimulation():
    def __init__(self,
                 cfg_file,
                 output_dir,
                 tl_ids,
                 ):
        
        self.cfg_file=cfg_file
        self.output_dir=output_dir
        self.tl_ids=tl_ids

    def simulation(self,
                   epoch_index,
                   rep,
                   demand_scaling_factor=None):

        if demand_scaling_factor is not None:
            command=[
                    "sumo",
                    "-c", f"./{self.cfg_file}",
                    "--scale",f"{demand_scaling_factor}",
                    "--verbose","--seed", f"{rep}",
                    "--output-prefix", f"{self.output_dir}{epoch_index}_{rep}_"
                    ]
        else:
            command=[
                    "sumo",
                    "-c", f"./{self.cfg_file}",
                    "--verbose","--seed", f"{rep}",
                    "--output-prefix", f"{self.output_dir}{epoch_index}_{rep}_"
                    ]
        try:
            traci.start(command, numRetries=300)
            connection_started = True
            for tl in self.tl_ids:
                traci.trafficlight.setProgram(tl, "1")
                program = traci.trafficlight.getAllProgramLogics(tl)
                program_used=traci.trafficlight.getProgram(tl)

            while (
                    traci.simulation.getMinExpectedNumber() > 0
                    and traci.simulation.getTime() < 64800
                ):
                    print(
                        "Simulation time:", traci.simulation.getTime(),
                        "| Expected vehicles:", traci.simulation.getMinExpectedNumber()
                    )
                    traci.simulationStep()

        except Exception as e:
            print(f"Erreur SUMO/TraCI : {e}")
            traceback.print_exc()
            raise

        finally:
            if connection_started:
                try:
                    traci.close()
                    print("TraCI connection closed.", flush=True)
                except Exception as e:
                    print(f"Error while closing TraCI: {e}", flush=True)
                    raise

        df_sim=pd.read_xml(f"./Ingolstadt Network/{self.output_dir}{epoch_index}_{rep}_InTAS.travel_times.xml",iterparse={"edge":["id","sampledSeconds","traveltime",\
                                                                                                                   "density","occupancy","speed","waitingTime",\
                                                                                                                    "timeLoss","departed","arrived","entered","left",\
                                                                                                                        "flow"]}).rename(columns={"id":"edge_id"})
        if df_sim.shape[0]!=len(df_sim["edge_id"].unique()):
            raise ValueError('This code asssumes there is  a single sampling time interval in the edge travel time xml file. If an edge id appears multiple times it is because there are multiple sampling intervals')

        df_sim.to_csv(f"{self.output_dir}{epoch_index}_{rep}_InTAS.travel_times.csv")

        
net=sumolib.net.readNet("Ingolstadt Network/ingolstadt.net.xml")
tl_ids=[tl.getID() for tl in net.getTrafficLights()]
sumo_simulator=SumoSimulation(
    cfg_file="./Ingolstadt Network/InTAS_test.sumocfg",
    output_dir="gt_data/",
    tl_ids=tl_ids
    
)
sumo_simulator.simulation(epoch_index="test",rep=-1)  
    
```
